```python
import os
import zipfile
import glob
import pandas as pd

data_folder = "/content/review_dataset"
os.makedirs(data_folder, exist_ok=True)

uploaded_files = list(uploaded.keys())
csv_files = []

for file_name in uploaded_files:
    if zipfile.is_zipfile(file_name):
        with zipfile.ZipFile(file_name, "r") as zip_ref:
            zip_ref.extractall(data_folder)
        csv_files.extend(
            glob.glob(data_folder + "/**/*.csv", recursive=True)
        )
    elif file_name.lower().endswith(".csv"):
        csv_files.append(file_name)

if not csv_files:
    raise FileNotFoundError(
        "CSV file dorakaledu. ZIP lopala CSV undho check cheyyi."
    )

print("Available CSV files:")
for i, file_name in enumerate(csv_files):
    print(i, os.path.basename(file_name))
```

```python
csv_path = csv_files[0]

df = pd.read_csv(
    csv_path,
    nrows=100000,
    encoding="utf-8",
    on_bad_lines="skip"
)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

df.head()
```

```python
df.columns = df.columns.str.strip()

required_columns = ["Score", "Text"]

if not all(col in df.columns for col in required_columns):
    raise ValueError(
        "Dataset lo Score and Text columns levu. "
        "Column names check cheyyi."
    )

df = df.dropna(subset=["Score", "Text"]).copy()

df["Score"] = pd.to_numeric(df["Score"], errors="coerce")
df = df.dropna(subset=["Score"])

df["Text"] = df["Text"].astype(str).str.strip()
df = df[df["Text"] != ""]

if "Summary" in df.columns:
    df["Text"] = (
        df["Summary"].fillna("").astype(str) + " " + df["Text"]
    ).str.strip()

print("Rows after cleaning:", len(df))
df[["Score", "Text"]].head()
```

```python
def get_sentiment(score):
    if score <= 2:
        return "Negative"
    elif score == 3:
        return "Neutral"
    else:
        return "Positive"

df["Sentiment"] = df["Score"].apply(get_sentiment)

print(df["Sentiment"].value_counts())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="Sentiment",
    order=["Negative", "Neutral", "Positive"]
)

plt.title("Distribution of Customer Review Sentiments")
plt.xlabel("Sentiment")
plt.ylabel("Number of Reviews")
plt.tight_layout()
plt.show()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline

X = df["Text"]
y = df["Sentiment"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

model = Pipeline([
    ("tfidf", TfidfVectorizer(
        max_features=20000,
        stop_words="english",
        ngram_range=(1, 2),
        min_df=2
    )),
    ("classifier", LogisticRegression(
        max_iter=1000,
        class_weight="balanced"
    ))
])

model.fit(X_train, y_train)

print("Model training completed!")
print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

y_pred = model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)

print(f"Model Accuracy: {accuracy * 100:.2f}%")
print("\nClassification Report:")
print(classification_report(y_test, y_pred, zero_division=0))

ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred,
    labels=["Negative", "Neutral", "Positive"],
    xticks_rotation=30
)

plt.title("Confusion Matrix")
plt.tight_layout()
plt.show()
```

```python
reviews = [
    "This product is excellent and works perfectly!",
    "Very poor quality. I regret buying this product.",
    "The product is okay, nothing special."
]

predictions = model.predict(reviews)

for review, sentiment in zip(reviews, predictions):
    print("Review:", review)
    print("Predicted Sentiment:", sentiment)
    print("-" * 50)
```
